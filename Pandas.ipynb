Find out the bats man who hit six highest

```python
delivery[delivery["batsman_runs"] == 4].groupby("batsman")['total_runs'].count().sort_values(ascending = False).head(5)
```

```python
delivery[delivery["batsman"] == "V Kohli"].groupby("bowling_team")["batsman_runs"].sum().sort_values(ascending=False)
```

```python
def runs_against(bats_man):
    team = delivery[delivery["batsman"] == bats_man].groupby("bowling_team")["batsman_runs"].sum().sort_values(ascending=False).head(1)
    return team

print(runs_against("G Gambhir"))
```

#### isin function

```python
delivery
```

```python
death_over = delivery[delivery["over"] > 15] # --> 16 to 20 over death over
```

```python
selected_batsman = death_over.groupby("batsman")["batsman_runs"].count() 
selected_batsman.shape
```

```python
special_batsman = selected_batsman[selected_batsman > 200]
x = special_batsman.index.tolist()
```

```python
final = death_over[death_over["batsman"].isin(x)]
```

strike rate = (runs of a player / total ball played by that player) * 100

```python
runs = final.groupby("batsman")["batsman_runs"].sum()
```

```python
balls = final.groupby("batsman")["batsman_runs"].count()
```

```python
((runs/balls)*100).sort_values(ascending=False).head()
```
